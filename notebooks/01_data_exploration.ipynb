```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="Quality of Sleep",
    palette="Blues"
)

plt.title("Distribution of Quality of Sleep")
plt.xlabel("Quality of Sleep")
plt.ylabel("Count")

plt.savefig(
    "../figures/quality_of_sleep_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Most participants have a sleep quality score between 6 and 8. Very low sleep quality scores are relatively uncommon, indicating the majority report average to good sleep quality.

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Sleep Duration",
    palette="Set2"
)

plt.title("Sleep Duration by Gender")
plt.xlabel("Gender")
plt.ylabel("Sleep Duration (Hours)")

plt.savefig(
    "../figures/sleep_duration_by_gender.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Both genders exhibit similar median sleep durations. The spread of sleep duration appears comparable, with only minor variation between male and female participants.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Heart Rate",
    bins=10,
    kde=True,
    color="green"
)

plt.title("Distribution of Heart Rate")
plt.xlabel("Heart Rate")
plt.ylabel("Count")

plt.savefig(
    "../figures/heart_rate_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Heart rate values cluster around 70 bpm, suggesting most participants fall within a normal resting heart rate range.

```python
numeric_df = df.select_dtypes(include=["number"])

plt.figure(figsize=(10,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.savefig(
    "../figures/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

### Observations

Sleep Duration and Quality of Sleep show a strong positive correlation (`r = 0.88`), suggesting that participants who sleep longer generally report higher sleep quality.

Stress Level and Quality of Sleep show a very strong negative correlation (`r = -0.90`), indicating that higher stress levels are strongly associated with lower reported sleep quality.

Stress Level and Sleep Duration show a strong negative correlation (`r = -0.81`), suggesting that participants experiencing higher stress tend to sleep for fewer hours.

Physical Activity Level and Daily Steps show a strong positive correlation (`r = 0.77`), indicating that participants with higher physical activity levels generally record more daily steps.

Stress Level and Heart Rate show a moderately strong positive correlation (`r = 0.67`), suggesting that higher stress levels are associated with higher resting heart rates in this dataset.

### Note

These correlations describe associations in the dataset and do not establish causation.

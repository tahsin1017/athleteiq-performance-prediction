### Sleep Disorder Investigation

The `Sleep Disorder` variable contains 219 missing values, compared with 78 participants with Sleep Apnea and 77 with Insomnia.

The average Quality of Sleep differs across these groups. Participants with Insomnia have an average sleep-quality score of 6.53, while participants with Sleep Apnea have an average of 7.21. Participants with missing Sleep Disorder values have the highest average score at 7.63.

The large number of missing values and their different sleep-quality distribution suggest that the missing values may represent participants without a recorded sleep disorder. However, this assumption should be verified before converting the missing values into a separate category.

```python
# Split Blood Pressure into Systolic and Diastolic values

df[["Systolic BP", "Diastolic BP"]] = df["Blood Pressure"].str.split(
    "/",
    expand=True
)

# Convert them from strings to numbers
df["Systolic BP"] = pd.to_numeric(df["Systolic BP"])
df["Diastolic BP"] = pd.to_numeric(df["Diastolic BP"])

# Check the result
df[["Blood Pressure", "Systolic BP", "Diastolic BP"]].head()
```

### Blood Pressure Feature Engineering

The `Blood Pressure` variable was originally stored as a combined string containing systolic and diastolic measurements.

The variable was split into two separate numerical features: `Systolic BP` and `Diastolic BP`. This makes the blood pressure information suitable for numerical analysis and potential use in machine learning models.

The original `Blood Pressure` column will be excluded from the final modeling features because its information is now represented by the two numerical variables.


```python
df[["Systolic BP", "Diastolic BP"]].describe()
```

### Blood Pressure Observations

The average systolic blood pressure is 128.55, while the average diastolic blood pressure is 84.65.

Systolic blood pressure ranges from 115 to 142, while diastolic blood pressure ranges from 75 to 95.

Both blood pressure variables show weak negative correlations with Quality of Sleep. Systolic BP has a correlation of -0.12, while Diastolic BP has a correlation of -0.11.

Systolic and Diastolic BP have a very strong positive correlation (r = 0.97), indicating that the two measurements are highly related.

Overall, blood pressure appears to have a relatively weak linear relationship with Quality of Sleep compared with variables such as Sleep Duration and Stress Level.

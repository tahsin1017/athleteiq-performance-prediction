# AthleteIQ Data Preprocessing

## Objective

Prepare the sleep and lifestyle dataset for machine learning by cleaning the data, handling missing values, transforming variables, and creating a processed dataset.

### Preprocessing Tasks

- Load the raw dataset
- Inspect the dataset
- Remove unnecessary columns
- Split Blood Pressure into numeric features
- Standardize BMI categories
- Handle missing values
- Check for duplicates
- Save the processed dataset

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/Sleep_health_and_lifestyle_dataset.csv")

df.head()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

```python
# Create a copy of the raw dataset for preprocessing
df_processed = df.copy()

df_processed.head()
```

```python
# Remove the identifier column
df_processed = df_processed.drop(columns=["Person ID"])

df_processed.head()
```

```python
print(df_processed["BMI Category"].value_counts())
```

```python
# Standardize BMI category names
df_processed["BMI Category"] = df_processed["BMI Category"].replace(
    {"Normal Weight": "Normal"}
)

print(df_processed["BMI Category"].value_counts())
```

```python
# Split Blood Pressure into two numeric columns
df_processed[["Systolic BP", "Diastolic BP"]] = (
    df_processed["Blood Pressure"]
    .str.split("/", expand=True)
    .astype(int)
)

df_processed[["Blood Pressure", "Systolic BP", "Diastolic BP"]].head()
```

```python
# Remove the original Blood Pressure column
df_processed = df_processed.drop(columns=["Blood Pressure"])

df_processed.head()
```

```python
print(df_processed.isnull().sum())
```

```python
# Replace missing sleep disorder values with "None"
df_processed["Sleep Disorder"] = df_processed["Sleep Disorder"].fillna("None")

print(df_processed["Sleep Disorder"].value_counts())
```

```python
print(df_processed.isnull().sum())
```

```python
print("Number of duplicate rows:", df_processed.duplicated().sum())
```

```python
print("Final dataset shape:", df_processed.shape)

df_processed.info()
```

```python
df_processed.head()
```

## Preprocessing Summary

The dataset was prepared for machine learning through the following steps:

- Removed `Person ID` because it is only an identifier and does not provide useful predictive information.
- Standardized the `BMI Category` values by combining `Normal Weight` with `Normal`.
- Split `Blood Pressure` into two numerical variables: `Systolic BP` and `Diastolic BP`.
- Removed the original `Blood Pressure` text column.
- Replaced missing values in `Sleep Disorder` with `None`.
- Checked the dataset for remaining missing values.
- Checked for duplicate records.
- Created a cleaned dataset ready for the next stage of feature preparation and machine learning.

```python
# Save the processed dataset
df_processed.to_csv(
    "../data/processed/processed_sleep_health_dataset.csv",
    index=False
)

print("Processed dataset saved successfully.")
```
